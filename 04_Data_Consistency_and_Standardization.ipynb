# Data Consistency and Standardization

Real-world datasets often contain inconsistencies such as different text formats, extra spaces, special characters, and features with different numerical scales.

Data consistency ensures that all data follows a uniform format, while standardization and scaling prepare numerical features for Machine Learning algorithms.

In this section, we will perform:

- Text Data Cleaning
- Numerical Data Scaling
- Feature Transformation

# Text Data Cleaning

Text data frequently contains inconsistencies such as uppercase and lowercase letters, extra spaces, special characters, and spelling mistakes.

Cleaning text data improves data quality and ensures consistency throughout the dataset.

## Lowercase Conversion

Converting text to lowercase ensures that values with different letter cases are treated as the same value.

Example:

- "Male"
- "male"
- "MALE"

After conversion:

- "male"

```python
import pandas as pd 
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
df = sns.load_dataset("titanic")
```

```python
df.head()
```

```python
df.columns
```

```python
df["Class"] = df["class"].str.lower()
df["Class"].head()
```

```python
df.head()
```

## Remove Extra Whitespaces

Leading and trailing spaces may cause duplicate values during analysis.

The `strip()` function removes unnecessary spaces.

- Example : "  Alisha Sajjad  " -> "Alisha Sajjad"

```python
df["Class"] = df["class"].str.strip()

df["Class"].head()
```

## Remove Special Characters

Special characters are often unnecessary for Machine Learning tasks.

Removing them creates cleaner and more consistent text data.

- Example -> "alisha@sajjad" → "alishasajjad"

```python
import re 

df["Class"] = df["Class"].apply(lambda x: re.sub(r"[^a-zA-Z\s]", "", x))

df["Class"].head()
```

## Unicode Normalization

Unicode normalization converts accented characters into standard English characters.

Example:

- café → cafe
- résumé → resume

```python
import unicodedata

def normalize_text(text):
    return unicodedata.normalize("NFKD", text).encode("ascii", "ignore").decode("utf-8")

df["Class"] = df["Class"].apply(normalize_text)

df["Class"].head()
```

## Spelling Correction

Some NLP applications require correcting spelling mistakes before analysis.

This technique is mainly used in Natural Language Processing (NLP).

```python
%pip install textblob
```

```python
from textblob import TextBlob 

df["Class"] = df["class"].apply(lambda x: str(TextBlob(x).correct()))

df["Class"].head()
```

# Numerical Data Standardization

Many Machine Learning algorithms perform better when numerical features are on a similar scale.

Scaling prevents features with larger values from dominating those with smaller values.

## Min-Max Scaling

Min-Max Scaling transforms numerical values into a range between **0 and 1**.

It is useful when the data does not contain significant outliers.

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df["Fare_MinMax"] = scaler.fit_transform(df[["fare"]])

df[["fare", "Fare_MinMax"]].head()
```

## Standard Scaling (Z-Score)

Standard Scaling transforms data so that:

- Mean = 0
- Standard Deviation = 1

It is commonly used when the data approximately follows a normal distribution.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df["Fare_Standard"] = scaler.fit_transform(df[["fare"]])

df[["fare", "Fare_Standard"]].head()
```

## Robust Scaling

Robust Scaling uses the median and interquartile range (IQR) instead of the mean and standard deviation.

It is particularly useful when the dataset contains many outliers.

```python
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

df["Fare_Robust"] = scaler.fit_transform(df[["fare"]])

df[["fare", "Fare_Robust"]].head()
```

## Log Transformation

Log Transformation reduces skewness and minimizes the influence of large values.

It is commonly applied to highly skewed numerical features.

```python
import numpy as np 

df["Fare_Log"] = np.log1p(df["fare"])

df[["fare", "Fare_Log"]].head()
```

## Power Transformation

Power Transformation makes data more normally distributed and reduces skewness.

It is useful for improving the performance of certain Machine Learning algorithms.

```python
from sklearn.preprocessing import PowerTransformer

scaler = PowerTransformer()

df["Fare_Power"] = scaler.fit_transform(df[["fare"]])

df[["fare", "Fare_Power"]].head()
```

## 📌 Key Points to Remember

- **Text Cleaning:** Converting text to lowercase, removing extra spaces, and correcting spelling mistakes help maintain consistency in textual data.

- **Numeric Scaling:** Standardization and transformation techniques normalize feature ranges and improve data distribution for Machine Learning algorithms.

- **Model Compatibility:** Clean, consistent, and properly scaled data enables Machine Learning models to learn more effectively, resulting in better performance and more accurate predictions.
