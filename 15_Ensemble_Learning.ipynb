# Ensemble Learning

Ensemble Learning is a machine learning technique that combines multiple models to produce a more accurate, stable, and robust prediction than any single model.

The main idea is based on the concept:

> **"Wisdom of Crowds"**

Just as a group of people often makes better decisions than a single individual, combining multiple machine learning models usually produces better predictions than relying on one model alone.

# Key Concepts of Ensemble Learning

## 1. Diversity of Models

Different machine learning models learn patterns from data in different ways.

Each model may make different predictions.

Combining these predictions often reduces overall errors and improves performance.

---

## 2. Wisdom of Crowds

Instead of relying on one model, Ensemble Learning combines predictions from multiple models.

The final prediction is usually:

- More accurate
- More stable
- Less sensitive to noise

# Types of Ensemble Learning

There are three major Ensemble Learning techniques:

1. Bagging (Bootstrap Aggregating)

2. Boosting

3. Stacking

Each method combines models differently to improve prediction performance.

# 1. Bagging (Bootstrap Aggregating)

Bagging trains multiple models independently using different random subsets of the training dataset.

The predictions from all models are then combined.

For Classification:
- Majority Voting

For Regression:
- Average Prediction

## Advantages

- Reduces variance
- Prevents overfitting
- Produces stable models

## Popular Algorithm

- Random Forest

# 2. Boosting

Boosting trains models sequentially.

Each new model focuses on correcting the mistakes made by the previous model.

The final model becomes stronger by learning from previous errors.

## Advantages

- High prediction accuracy
- Reduces bias
- Works well on complex datasets

# Popular Boosting Algorithms

## 2.1 AdaBoost (Adaptive Boosting)

AdaBoost trains weak learners one after another.

Each new learner gives more attention to the samples that were incorrectly classified by the previous learner.

### Best For

- Classification problems

### Advantages

- Simple
- Fast
- Improves weak models

## 2.2 Gradient Boosting Machine (GBM)

Gradient Boosting improves predictions by minimizing the errors (loss function) using gradient optimization.

Each new model tries to reduce the residual errors of the previous model.

### Best For

- Regression
- Classification

### Advantages

- High accuracy
- Flexible

## 2.3 XGBoost (Extreme Gradient Boosting)

XGBoost is an optimized version of Gradient Boosting.

It is faster, memory-efficient, and includes regularization to reduce overfitting.

### Features

- Fast training
- Regularization
- Parallel processing
- Excellent performance

### Best For

- Large datasets
- Machine Learning competitions

## 2.4 LightGBM (Light Gradient Boosting Machine)

LightGBM is a lightweight and high-speed boosting algorithm.

It uses a leaf-wise tree growth strategy, making training much faster than traditional Gradient Boosting.

### Features

- Very fast
- Low memory usage
- Efficient for large datasets

### Best For

- Large datasets
- High-dimensional data

## 2.5 CatBoost

CatBoost is a boosting algorithm specially designed to work with categorical features.

It can process categorical variables directly without extensive preprocessing.

### Features

- Handles categorical data
- High accuracy
- Reduces overfitting

### Best For

- Datasets with many categorical variables

## 2.6 Stochastic Gradient Boosting

Stochastic Gradient Boosting is a variation of Gradient Boosting.

Instead of using the entire dataset for every iteration, it randomly selects a subset of training data.

This randomness improves generalization and reduces overfitting.

### Best For

- Large noisy datasets

# Summary of Boosting Algorithms

| Algorithm | Main Strength | Best Use |
|------------|---------------|-----------|
| AdaBoost | Corrects previous mistakes | Classification |
| Gradient Boosting | Optimizes prediction errors | Regression & Classification |
| XGBoost | Fast and regularized | Large datasets |
| LightGBM | High speed and low memory | Large datasets |
| CatBoost | Handles categorical data | Mixed datasets |
| Stochastic Gradient Boosting | Adds randomness | Noisy datasets |

# 3. Stacking

Stacking combines predictions from multiple different machine learning models.

A final model, called the **Meta Model**, learns from these predictions and produces the final output.

Example:

Base Models

- Decision Tree
- Logistic Regression
- Support Vector Machine (SVM)

↓

Meta Model

↓

Final Prediction

## Advantages

- Combines strengths of different algorithms
- Often produces higher accuracy

# 4. Voting Ensemble

Voting combines predictions from multiple classifiers.

There are two types of Voting:

1. Hard Voting
2. Soft Voting

## Hard Voting

Hard Voting selects the class predicted by the majority of models.

Every model has equal importance.

### Example

Model 1 → Class A

Model 2 → Class A

Model 3 → Class B

Final Prediction:

**Class A** (Majority Vote)

## Soft Voting

Soft Voting uses prediction probabilities instead of class labels.

Each model contributes according to its confidence.

The class with the highest combined probability becomes the final prediction.

### Example

Model 1

Class A → 0.90

Class B → 0.10

Model 2

Class A → 0.70

Class B → 0.30

Final Prediction:

**Class A**

# When to Use Hard Voting and Soft Voting

## Hard Voting

Use Hard Voting when:

- Models only provide class labels.
- Every model has equal importance.

---

## Soft Voting

Use Soft Voting when:

- Models provide prediction probabilities.
- Confidence scores are available.
- More reliable predictions are required.

Ensemble Learning improves prediction performance by combining multiple machine learning models. Techniques such as Bagging, Boosting, Stacking, and Voting help reduce overfitting, improve generalization, and produce more accurate and reliable predictions than a single model.
