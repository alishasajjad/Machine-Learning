# 1. Introduction

## 🎯 Objective

Data Splitting is the process of dividing a dataset into different subsets before training a Machine Learning model.

Instead of training and testing the model on the same data, we separate the dataset into different parts to evaluate how well the model performs on unseen data.

Proper data splitting helps prevent overfitting, provides a fair evaluation of the model, and improves its ability to generalize.

### Why Data Splitting?

- Train the model on one dataset and evaluate it on another.
- Prevent overfitting.
- Measure real-world model performance.
- Compare different Machine Learning models fairly.
- Build more reliable and robust models.

# 2. Train-Test Split

## What is Train-Test Split?

Train-Test Split is the simplest way to divide a dataset.

The dataset is split into two parts:

- **Training Set** → Used to train the model.
- **Testing Set** → Used to evaluate the trained model.

A common split ratio is:

- 80% Training
- 20% Testing

````python
import numpy as np 
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
df = sns.load_dataset("titanic")
````

````python
from sklearn.model_selection import train_test_split 

# Features (Independent Variables)
X = df.drop(columns=["survived"])

# Target (Dependent Variable)
Y = df["survived"]

# Split dataset into Training (80%) and Testing (20%)
X_train, X_test, Y_train, Y_tes = train_test_split(X,Y,test_size = 0.2, random_state=42)

print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
````

### Explanation

- **test_size=0.2** means 20% of the dataset is used for testing.
- **random_state=42** ensures reproducible results.
- The remaining 80% of the data is used for training.

# 3. Train-Validation-Test Split

## What is Train-Validation-Test Split?

Sometimes a single test set is not enough.

The dataset is divided into three parts:

- **Training Set** → Used to train the model.
- **Validation Set** → Used to tune hyperparameters.
- **Testing Set** → Used for final evaluation.

A common split is:

- 60% Training
- 20% Validation
- 20% Testing

````python
from sklearn.model_selection import train_test_split

X = df.drop(columns=["survived"]) # Features
Y = df["survived"]  # Target

# First split: 60% Training, 40% Temporary Data
X_train, X_temp, Y_train, Y_temp = train_test_split(X,Y, test_size = 0.2, random_state=42)

# Second split:
# Temporary data into Validation and Testing
X_val, X_test, Y_val, Y_test = train_test_split(X_temp,Y_temp,test_size=0.5,random_state=42)

print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Testing:", X_test.shape)
````

### Explanation

The data is divided in two steps:

1. First split into **Training (60%)** and **Temporary (40%)**.
2. The temporary data is then divided equally into **Validation (20%)** and **Testing (20%)**.

# 4. Stratified Sampling

## What is Stratified Sampling?

When the target variable is imbalanced, a random split may create unequal class distributions.

Stratified Sampling ensures that each subset contains nearly the same class distribution as the original dataset.

This technique is commonly used in classification problems.

````python
from sklearn.model_selection import train_test_split

X = df.drop(columns=["survived"]) # Features
Y = df["survived"] # Target

# Perform stratified train-test split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.5, stratify=Y, random_state=42)

print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
````

````python
# Display class distribution
print("Training Class Distribution")
print(Y_train.value_counts(normalize=True))

print("\nTesting Class Distribution")
print(Y_test.value_counts(normalize=True))
````

### Explanation

Using

```python
stratify=y
```

ensures that both the training and testing datasets maintain approximately the same class distribution as the original dataset.

# 5. Cross Validation

## What is Cross Validation?

Cross Validation is a model evaluation technique.

Instead of using a single train-test split, the dataset is divided into multiple folds.

The model is trained and tested multiple times, and the average performance is calculated.

The most common method is **5-Fold Cross Validation**.

````python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier
import pandas as pd

X = df.drop(columns=["survived", "alive"]) # Features
Y = df["survived"] # Target

# Fill missing values
X["age"] = X["age"].fillna(X["age"].median())
X["fare"] = X["fare"].fillna(X["fare"].median())
X["embarked"] = X["embarked"].fillna(X["embarked"].mode()[0])
X["embark_town"] = X["embark_town"].fillna(X["embark_town"].mode()[0])
X["deck"] = X["deck"].fillna(X["deck"].mode()[0])

# One-Hot Encoding
X = pd.get_dummies(X, drop_first=True)

# Create Random Forest model
model = RandomForestClassifier(random_state=42)

# Perform 5-Fold Cross Validation
scores = cross_val_score(model, X, Y, cv=5)

print("Cross Validation Scores:")
print(scores)

print("\nAverage Score:")
print(scores.mean())
````

### Why Cross Validation?

- Uses the entire dataset efficiently.
- Produces a more reliable evaluation.
- Reduces the risk of overfitting.
- Helps compare different models.
- Widely used in Machine Learning projects.
