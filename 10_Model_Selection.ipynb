# Model Selection

## 🎯 Objective

Model Selection is the process of choosing the most suitable Machine Learning algorithm for a specific problem.

Selecting the right model is important because different algorithms perform differently depending on the type of data, problem, and evaluation metric.

Before selecting a model, we should understand:

- Classification
- Regression
- Evaluation Metrics
- Types of Machine Learning Models

This notebook introduces the basic concepts required before implementing Machine Learning algorithms.

# 1. Classification

## What is Classification?

Classification is a type of **Supervised Machine Learning** in which the model learns from labeled data and predicts a **categorical output**.

The goal is to assign an input to one of the predefined classes.

### Examples

- Email Spam Detection (Spam / Not Spam)
- Disease Diagnosis (Healthy / Diseased)
- Image Recognition (Cat / Dog / Elephant)

Classification predicts **labels or categories**, not numerical values.

## 1.1 Types of Classification

### Binary Classification

Binary Classification contains only **two classes**.

#### Examples

- Spam / Not Spam
- Fraud / Legitimate
- Pass / Fail
- Healthy / Diseased

---

### Multiclass Classification

Multiclass Classification contains **three or more classes**.

#### Examples

- Digit Recognition (0–9)
- Animal Classification
- Flower Species Classification
- Image Recognition

# 2. Regression

## What is Regression?

Regression is a type of **Supervised Machine Learning** used to predict **continuous numerical values**.

Instead of predicting categories, regression predicts numbers.

### Examples

- House Price Prediction
- Stock Price Prediction
- Temperature Prediction
- Sales Forecasting

Regression is mainly used whenever the target variable is numerical.

# 3. Evaluation Metrics

## Why Do We Need Evaluation Metrics?

After training a Machine Learning model, we need to measure how well it performs.

Evaluation metrics help us compare different models and choose the best one.

## 3.1 Classification Metrics

Classification models are evaluated using different performance metrics.

### 3.1.1 Accuracy

Accuracy measures the percentage of correct predictions made by the model.

**Formula**

Accuracy = Correct Predictions / Total Predictions

### Interpretation

- High Accuracy → More correct predictions
- Low Accuracy → More incorrect predictions

> **Note:** Accuracy may be misleading when the dataset is imbalanced.

### 3.1.2 Precision

Precision measures how many predicted positive cases are actually positive.

**Formula**

Precision = TP / (TP + FP)

### Interpretation

- High Precision means fewer False Positives.
- Useful for spam detection and recommendation systems.

### 3.1.3 Recall (Sensitivity)

Recall measures how many actual positive cases are correctly identified.

**Formula**

Recall = TP / (TP + FN)

### Interpretation

- High Recall means fewer False Negatives.
- Important in medical diagnosis and fraud detection.

### 3.1.4 F1 Score

F1 Score is the harmonic mean of Precision and Recall.

It provides a balanced evaluation when both Precision and Recall are important.

**Formula**

F1 Score = 2 × (Precision × Recall) / (Precision + Recall)

### Interpretation

Higher F1 Score indicates better overall classification performance.

### 3.1.5 Confusion Matrix

A Confusion Matrix compares predicted values with actual values.

It contains four important values:

- True Positive (TP)
- True Negative (TN)
- False Positive (FP)
- False Negative (FN)

The confusion matrix provides a detailed view of model performance beyond accuracy.

### 3.1.6 ROC Curve and AUC

### ROC Curve

The Receiver Operating Characteristic (ROC) Curve shows the relationship between:

- True Positive Rate (TPR)
- False Positive Rate (FPR)

---

### AUC (Area Under the Curve)

AUC measures the overall performance of a classification model.

- AUC = 1 → Perfect model
- AUC = 0.5 → Random guessing
- AUC < 0.5 → Poor model

## 3.2 Regression Metrics

Regression models are evaluated using numerical error measurements.

### 3.2.1 Mean Absolute Error (MAE)

MAE calculates the average absolute difference between actual and predicted values.

Lower MAE indicates better performance.

### 3.2.2 Mean Squared Error (MSE)

MSE calculates the average of squared prediction errors.

Large errors receive a higher penalty because of squaring.

### 3.2.3 Root Mean Squared Error (RMSE)

RMSE is the square root of MSE.

It is easier to interpret because it has the same unit as the target variable.

Lower RMSE indicates better performance.

### 3.2.4 R-Squared (R² Score)

R² measures how much variance in the target variable is explained by the model.

### Interpretation

- R² = 1 → Perfect prediction
- R² = 0 → Model explains no variance

Higher R² indicates a better regression model.
